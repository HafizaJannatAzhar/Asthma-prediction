```python
import pandas as pd
import joblib

lr_model = joblib.load('lr_model.pkl')
rf_model = joblib.load('rf_model.pkl')
scaler = joblib.load('scaler.pkl')

THRESHOLD = 0.3

manual_data = {
    'Gender': 0,
    'Smoking': 0,
    'PhysicalActivity': 0,
    'DietQuality': 1,
    'SleepQuality': 1,
    'PollutionExposure': 1,
    'PollenExposure': 1,
    'DustExposure': 0,
    'PetAllergy': 0,
    'FamilyHistoryAsthma': 0,
    'HistoryOfAllergies': 0,
    'Eczema': 0,
    'HayFever': 0,
    'GastroesophagealReflux': 0,
    'LungFunctionFEV1': 0,
    'LungFunctionFVC': 0,
    'Wheezing': 1,
    'ShortnessOfBreath': 0,
    'ChestTightness': 0,
    'Coughing': 0,
    'NighttimeSymptoms': 0,
    'ExerciseInduced': 1,
    'Age_Teenager': 0,
    'Age_Adult': 0,
    'Age_Senior': 1,
    'BMI_Underweight': 0,
    'BMI_Normal': 0,
    'BMI_Overweight': 1,
    'BMI_Obese': 0,
    'Ethnicity_Caucasian': 0,
    'Ethnicity_AfricanAmerican': 1,
    'Ethnicity_Asian': 0,
    'Ethnicity_Other': 0,
    'Education_None': 0,
    'Education_HighSchool': 0,
    'Education_Bachelors': 0,
    'Education_Higher': 1
}

manual_df = pd.DataFrame([manual_data])

# ---------- Random Forest ----------
rf_proba = rf_model.predict_proba(manual_df)[0][1]
rf_pred = 1 if rf_proba >= THRESHOLD else 0

print("=== Random Forest ===")
print("Prediction:", "Asthma" if rf_pred == 1 else "No Asthma")
print(f"Probability: {rf_proba:.2%}")

# ---------- Logistic Regression ----------
manual_scaled = scaler.transform(manual_df)
lr_proba = lr_model.predict_proba(manual_scaled)[0][1]
lr_pred = 1 if lr_proba >= THRESHOLD else 0

print("\n=== Logistic Regression ===")
print("Prediction:", "Asthma" if lr_pred == 1 else "No Asthma")
print(f"Probability: {lr_proba:.2%}")
```
